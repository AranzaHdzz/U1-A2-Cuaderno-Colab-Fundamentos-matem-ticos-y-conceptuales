### 1. Uso de las funciones piso (*floor*) y techo (*ceiling*) en computación

* **Definición fundamental:** Para cualquier número real \\(x\\), la función **piso** (\\(\lfloor x \rfloor\\)) obtiene el entero más grande que es menor o igual a \\(x\\), mientras que la función **techo** (\\(\lceil x \rceil\\)) obtiene el entero más pequeño que es mayor o igual a \\(x\\).
* **Codificación y tamaño de representación:** Una aplicación directa en ciencias de la computación consiste en determinar el **número mínimo de bits** necesarios para codificar o almacenar una colección de \\(n\\) objetos, la cual se calcula mediante la función techo del logaritmo en base 2: \\(\lceil \log_2 n \rceil\\).
* **División discreta de entradas:** Al trabajar con algoritmos recursivos o estructuras de datos que dividen el tamaño del problema \\(n\\) a la mitad, las funciones piso y techo garantizan la consistencia aritmética exacta mediante la relación \\(\lceil n/2 \rceil + \lfloor n/2 \rfloor = n\\).
* **Límites y propiedades:** Sirven para establecer cotas matemáticas sobre valores continuos mediante la desigualdad \\(x - 1 < \lfloor x \rfloor \le x \le \lceil x \rceil < x + 1\\), además de ser funciones monótonas crecientes.

---

### 2. Por qué la función exponencial crece más rápido que cualquier polinomio

El texto establece formalmente que cualquier función **exponencial positiva** con base mayor a 1 (\\(a^n\\) con \\(a > 1\\)) crece a un ritmo superior al de cualquier **función polinomial** (\\(n^b\\)) debido a las siguientes razones formales:

* **Demostración mediante el límite al infinito:** La comparación directa entre la tasa de crecimiento de un polinomio \\(n^b\\) y una función exponencial \\(a^n\\) se fundamenta en el valor de su límite cuando el tamaño de la entrada \\(n\\) tiende al infinito:
  \\[\lim_{n \to \infty} \frac{n^b}{a^n} = 0\\]
* **Relación de notación asintótica (\\(o\\) pequeña):** Al dar cero este límite, se concluye formalmente la relación \\(n^b = o(a^n)\\) para cualquier constante \\(b\\) y para cualquier base \\(a > 1\\).
* **Dominancia asintótica:** Esta propiedad demuestra que las potencias de \\(n\\) crecen más lentamente que cualquier función exponencial. Como consecuencia, conforme \\(n\\) aumenta, el tiempo de ejecución de un algoritmo exponencial se vuelve prohibitivamente costoso y supera a cualquier complejidad polinomial \\(O(n^k)\\).

**RESPUESTA:**
### Definición y Características de un Algoritmo

Un **algoritmo** es un conjunto ordenado y finito de instrucciones precisas y no ambiguas que transforma un conjunto de datos de entrada en un resultado o salida para resolver un problema específico.

Sus **características clave** son:
* **Entrada y Salida:** Define claramente los datos iniciales que recibe y la solución que debe producir.
* **Preciso / No ambiguo:** Cada paso está claramente especificado y tiene un orden de ejecución exacto sin confusión.
* **Finito:** Finaliza tras ejecutar un número determinado y finito de pasos para cualquier entrada.
* **Determinista (Correcto):** Produce siempre el mismo resultado ante el mismo conjunto de datos de entrada.
* **Efectivo:** Cada instrucción es realizable en un tiempo y con recursos finitos.

---

### El Problema del Paro (*Halting Problem*)

Formulado por **Alan Turing** en 1936, el **Problema del Paro** consiste en determinar si existe un algoritmo general capaz de analizar cualquier programa arbitrario y sus datos de entrada para predecir si este se **detendrá eventualmente** o si continuará ejecutándose en un **ciclo infinito**.

Turing demostró que este es un **problema indecidible**, lo que significa que **es imposible construir un programa universal** que resuelva con éxito este problema para todas las combinaciones posibles de programas y datos.

---

### Análisis a Priori vs. Prueba a Posteriori

La diferencia principal radica en el momento de la evaluación y en la dependencia del entorno de ejecución:

* **Análisis a Priori (Teórico):** Se lleva a cabo durante la **fase de diseño**, antes de escribir o ejecutar el código. Evalúa la eficiencia mediante una función matemática o el crecimiento asintótico (como el tiempo de ejecución \\(T(n)\\)), permitiendo predecir el costo del algoritmo de forma **independiente del hardware, software o lenguaje de programación**.
* **Prueba a Posteriori (Empírica):** Se realiza **después de codificar** el algoritmo en un lenguaje de programación. Consiste en ejecutar el programa en una computadora real con datos de prueba seleccionados para medir empíricamente los **tiempos de reloj y el espacio de memoria** utilizados.


- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**
El comportamiento de un algoritmo ante distintas entradas de un mismo tamaño \\(n\\) se evalúa considerando tres escenarios principales:

* **Peor caso (*Worst case*):** Ocurre con la entrada de tamaño \\(n\\) que provoca que el algoritmo ejecute la **máxima cantidad de operaciones básicas** y consuma el mayor tiempo de ejecución. Este análisis proporciona un límite o **cota superior garantizada**, asegurando que el tiempo de ejecución nunca superará dicho valor para ninguna instancia del problema. Es la métrica más utilizada en la práctica por brindar certeza operativa y por ser generalmente más accesible de calcular que el caso promedio.

* **Mejor caso (*Best case*):** Corresponde a la entrada de tamaño \\(n\\) que requiere la **mínima cantidad de operaciones básicas** y permite completar la ejecución en el menor tiempo posible. Aunque define una cota inferior de tiempo, suele ser poco representativo para estimar el rendimiento habitual del algoritmo. No obstante, resulta útil en algoritmos como el ordenamiento por inserción, donde el rendimiento ante entradas ideales o casi ordenadas se mantiene sumamente alto.

* **Caso promedio (*Average case*):** Determina el número **esperado o típico de operaciones básicas** que el algoritmo ejecuta ante una entrada cualquiera de tamaño \\(n\\). Para su cálculo, es indispensable definir o asumir una **distribución de probabilidad** sobre el conjunto de entradas posibles. Aunque requiere un análisis matemático significativamente más complejo, ofrece la estimación más realista del desempeño en escenarios de uso cotidiano.

## 4. Sección Práctica 2: Iteración vs. Recursión
Un algoritmo recursivo debe llamarse a sí mismo resolviendo un problema menor e incluir siempre un "caso base" para evitar ciclos infinitos.

    else: return fibonacci_recursivo(n-1) + fibonacci_recursivo(n-2)

````python
# [CÓDIGO 2.1] Sucesión de Fibonacci Recursiva
# Fi = Fi-1 + Fi-2 para i >= 2; F0 = 0, F1 = 1

def fibonacci_recursivo(n: int) -> int:
    if n < 0:
        raise ValueError("n debe ser un entero mayor o igual a 0.")
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)


# Ejemplo de prueba
for i in range(10):
    print(f"F({i}) = {fibonacci_recursivo(i)}")
````


### [CÓDIGO 2.2] Factorial (Iterativo)

````python
# [CÓDIGO 2.2] Factorial (Iterativo)
# n! = 1 * 2 * 3 * ... * n (con 0! = 1)

def factorial_iterativo(n: int) -> int:
    if n < 0:
        raise ValueError("n debe ser un entero mayor o igual a 0.")
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado


# Ejemplo de prueba
for i in range(6):
    print(f"{i}! = {factorial_iterativo(i)}")
````

### Pregunta de Análisis 1 (Responde en una celda de texto):
Al analizar un algoritmo de ordenamiento sobre una lista, ¿qué escenarios representarían el "caso peor" y el "caso mejor" de los datos de entrada según la literatura revisada?

**RESPUESTA:**

### 1. Algoritmos de ordenamiento directo e iterativos (Burbuja, Inserción)
* **Peor caso:** Ocurre cuando la lista inicial está **ordenada en sentido inverso** (es decir, totalmente desordenada respecto al criterio solicitado). En este escenario, cada elemento requiere desplazarse por todo el arreglo, alcanzando un tiempo de ejecución de orden cuadrático **\\(O(n^2)\\)**.
* **Mejor caso:** Se presenta cuando la lista ya está **completamente ordenada** en la dirección deseada. En algoritmos adaptativos como Inserción o Burbuja, esto permite verificar los datos en una sola pasada realizando \\(n-1\\) comparaciones y cero intercambios, reduciendo la complejidad a un tiempo lineal **\\(O(n)\\)**.

---

### 2. Algoritmos basados en Divide y Vencerás (ej. Quicksort)
* La literatura destaca que en **Quicksort**, los escenarios no dependen únicamente de qué tan ordenada esté la lista inicial, sino del **balance de la partición generado por el pivote elegido**.
* **Peor caso:** Ocurre cuando el pivote genera subarreglos extremadamente desequilibrados (por ejemplo, al tomar el primer o último elemento como pivote en una lista que ya está ordenada o en orden inverso). En este caso, la recursión no divide el problema eficientemente y el tiempo se degrada a **\\(O(n^2)\\)**.
* **Mejor caso:** Ocurre cuando el pivote divide la lista exactamente por la mitad en dos sublistas de igual tamaño, logrando un orden de complejidad óptimo de **\\(O(n \log n)\\)**.

---

### 3. Algoritmos independientes del orden (ej. Mergesort, Heapsort)
* En algoritmos como **Mergesort** o **Heapsort**, la disposición u orden inicial de la lista no altera sustancialmente el proceso de división o de construcción del árbol/montículo. Por ello, se considera que **no presentan variaciones drásticas entre casos**, manteniendo un tiempo de ejecución de **\\(O(n \log n)\\)** tanto en el mejor como en el peor de los escenarios.

### Instrucciones:
1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
# 1. Factorial recursivo: n! = n * (n - 1)! (con 0! = 1 y 1! = 1)
def factorial_recursivo(n: int) -> int:
    if n < 0:
        raise ValueError("n debe ser un entero mayor o igual a 0.")
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursivo(n - 1)


# 2. Fibonacci iterativo: F0 = 0, F1 = 1; Fi = Fi-1 + Fi-2 para i >= 2
def fibonacci_iterativo(n: int) -> int:
    if n < 0:
        raise ValueError("n debe ser un entero mayor o igual a 0.")
    if n == 0:
        return 0
    if n == 1:
        return 1

    anterior_2 = 0  # F(0)
    anterior_1 = 1  # F(1)

    for _ in range(2, n + 1):
        actual = anterior_1 + anterior_2
        anterior_2 = anterior_1
        anterior_1 = actual

    return anterior_1


# Ejemplos de prueba
print("Factorial recursivo de 5:", factorial_recursivo(5))  # 120
print("Fibonacci iterativo de 10:", fibonacci_iterativo(10))  # 55
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time


def fibonacci_recursivo(n: int) -> int:
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)


def fibonacci_iterativo(n: int) -> int:
    if n == 0:
        return 0
    elif n == 1:
        return 1

    anterior_2 = 0
    anterior_1 = 1

    for _ in range(2, n + 1):
        actual = anterior_1 + anterior_2
        anterior_2 = anterior_1
        anterior_1 = actual

    return anterior_1


n = 35

# Medición Fibonacci recursivo
inicio_rec = time.time()
res_rec = fibonacci_recursivo(n)
fin_rec = time.time()
tiempo_rec = fin_rec - inicio_rec

# Medición Fibonacci iterativo
inicio_it = time.time()
res_it = fibonacci_iterativo(n)
fin_it = time.time()
tiempo_it = fin_it - inicio_it

print(f"Resultado recursivo ({n}): {res_rec} | Tiempo: {tiempo_rec:.6f} segundos")
print(f"Resultado iterativo ({n}): {res_it} | Tiempo: {tiempo_it:.8f} segundos")
````

3. Elabora un cuadro comparativo concluyendo por qué una de las dos crece de forma polinomial y la otra tiene tintes exponenciales.

### Cuadro Comparativo: Fibonacci Iterativo/Dinámico vs. Fibonacci Recursivo Directo

| Criterio | Versión Iterativa / Programación Dinámica | Versión Recursiva Directa |
| :--- | :--- | :--- |
| **Complejidad Temporal \\(T(n)\\)** | **Polinomial / Lineal (\\(O(n)\\))** | **Exponencial (\\(\Theta(\phi^n)\\) / \\(O(2^n)\\))** |
| **Estructura de Control** | Bucle secuencial acumulativo de \\(2\\) hasta \\(n\\). | Árbol de llamadas recursivas con bifurcación binaria. |
| **Tratamiento de Subproblemas** | Calcula cada valor de la serie **una sola vez** y lo almacena o reutiliza. | Recalcula los mismos valores **múltiples veces de forma redundante**. |
| **Ecuación de Recurrencia / Costo** | \\(T(n) = T(n-1) + O(1) \implies T(n) \in O(n)\\). | \\(T(n) = T(n-1) + T(n-2) + O(1) \implies T(n) \in O(2^n)\\). |
| **Comportamiento ante \\(n\\) grande** | **Eficiente y escalable:** procesar \\(n = 100\\) toma fracciones de milisegundo. | **Prohibitivo:** el número de operaciones explota exponencialmente. |

---

### Conclusión: ¿Por qué una crece polinomialmente y la otra exponencialmente?

1. **Crecimiento Polinomial (Versión Iterativa / Programación Dinámica):**
   * **Secuencialidad del trabajo:** Al incrementar el tamaño de entrada \\(n\\) en 1 unidad, el algoritmo iterativo únicamente añade **una iteración fija** al bucle.
   * **Inexistencia de redundancia:** Dado que el algoritmo resuelve cada término apoyándose en los dos inmediatos anteriores ya calculados, la cantidad total de operaciones suma \\(n\\) pasos simples. La función de tiempo \\(T(n)\\) responde a un polinomio de primer grado (\\(T(n) = a \cdot n + b\\)), manteniéndose dentro de la clase de complejidad lineal \\(O(n)\\).

2. **Tintes Exponenciales (Versión Recursiva Directa):**
   * **Ramificación geométrica del árbol:** La llamada recursiva no es lineal, sino doble (\\(F(n-1) + F(n-2)\\)). Esto transforma la ejecución en un **árbol de decisión binario**.
   * **Multiplicación del trabajo por nivel:** Al incrementar \\(n\\) en 1 unidad, no se agrega un solo paso, sino que se **duplican casi todos los nodos del árbol de llamadas**. Como no existen mecanismos de memoria, la cantidad total de llamadas recursivas crece de forma geométrica en proporción directa a la Razón Dorada (\\(\phi \approx 1.618\\)), haciendo que la complejidad escale a un orden \\(\Theta(\phi^n)\\) o \\(O(2^n)\\).
